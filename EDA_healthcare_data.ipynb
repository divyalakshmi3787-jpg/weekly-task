```python
# 17. Create a New Feature - Length of Stay

df["Length of Stay"] = (
    pd.to_datetime(df["Discharge Date"]) -
    pd.to_datetime(df["Date of Admission"])
).dt.days

df[["Date of Admission", "Discharge Date", "Length of Stay"]].head()
```

```python
# 18. Create a New Feature - Admission Year

df["Admission Year"] = pd.to_datetime(
    df["Date of Admission"]
).dt.year

df[["Date of Admission", "Admission Year"]].head()
```

```python
# 19. Create a New Feature - Admission Month

df["Admission Month"] = pd.to_datetime(
    df["Date of Admission"]
).dt.month

df[["Date of Admission", "Admission Month"]].head()
```

```python
df.groupby("Admission Year")["Billing Amount"].mean()
```

```python
monthly_admissions = (
    df.groupby("Admission Month")["Name"]
    .count()
    .reset_index()
)

monthly_admissions
```

```python
plt.figure(figsize=(9, 5))

sns.lineplot(
    data=monthly_admissions,
    x="Admission Month",
    y="Name",
    marker="o"
)

plt.title("Monthly Number of Admissions")
plt.xlabel("Admission Month")
plt.ylabel("Number of Admissions")

plt.show()
```

```python
# 1. BAR CHART
# Patients by Medical Condition
# =========================================

condition_count = df["Medical Condition"].value_counts()

plt.figure(figsize=(8,5))
plt.bar(condition_count.index, condition_count.values)

plt.title("Patients by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)

plt.show()
```

```python
# 2. LINE CHART
df.groupby("Medical Condition")["Billing Amount"].mean().plot(kind="line", marker="o")

plt.title("Average Billing Amount")
plt.xlabel("Medical Condition")
plt.ylabel("Billing Amount")
plt.show()
```

```python
# 3. PIE CHART
df["Gender"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.title("Gender Distribution")
plt.ylabel("")

plt.show()
```

```python
# 4. Scatter Plot

plt.scatter(df["Age"], df["Room Number"])

plt.title("Age vs Room Number")
plt.xlabel("Age")
plt.ylabel("Room Number")

plt.show()
```

```python
# 5. BOX PLOT

sns.boxplot(x="Admission Type", y="Billing Amount", data=df)

plt.title("Billing Amount by Admission Type")

plt.show()
```

REPORT

# The Healthcare Dataset contains 55,500 rows and 15 columns, covering patient information, medical conditions, admission details, billing amounts, insurance providers, and length of hospital stay. The data was cleaned and prepared for analysis, and date columns were converted into the appropriate format. Exploratory Data Analysis (EDA) was performed using bar charts, pie charts, line charts, boxplots, scatter plots, and correlation analysis to understand patient demographics, admission patterns, billing amounts, and length of stay. The analysis showed that the patients' ages ranged from 13 to 89 years, with an average age of approximately 51.5 years. Male patients had a slightly higher representation, and Arthritis was one of the most common medical conditions. Elective admission was the most frequent admission type, while Emergency and Urgent admissions were also significant. The average billing amount was approximately $25,539.32, and Obesity recorded one of the higher average billing amounts among the medical conditions. The average length of hospital stay was approximately 15.5 days, with only a very weak relationship between length of stay and billing amount. August recorded the highest number of admissions, while February recorded the lowest number of admissions. Correlation analysis showed very weak relationships between age, billing amount, and length of stay. The analysis highlights the importance of monitoring common medical conditions, planning hospital resources according to admission patterns, and managing billing and patient stay effectively. Hospitals can use these data-driven insights to improve resource planning, monitor patient admissions, and support better healthcare management decisions.
